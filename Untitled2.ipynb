```python
# ==================== 完整修复版训练脚本（兼容低算力 GPU） ====================
import json
import torch
from datasets import Dataset
from transformers import AutoModelForCausalLM, AutoTokenizer
from peft import LoraConfig
from trl import SFTTrainer, SFTConfig

# ================= 1. 路径与模型 =================
DATA_PATH = "social_diagnosis_10k.json"
MODEL_NAME = "Qwen/Qwen3-1.7B"
OUTPUT_DIR = "./qwen1.7b-social-diagnosis-lora"

# ================= 2. 读取并转换为 messages =================
with open(DATA_PATH, "r", encoding="utf-8") as f:
    raw_data = json.load(f)

SYSTEM_PROMPT = """你是一个中文社交关系诊断助手。
请根据用户提供的聊天记录，输出严格的 JSON 对象，必须包含以下字段：
- emotion: 字符串
- sentiment_polarity: 字符串
- affection_score: 整数
- trend: 字符串
- risk_level: 字符串
- risk_and_subtext: 字符串

不要输出 JSON 以外的任何解释、前后缀或 Markdown 代码块。"""

def convert_to_messages(example):
    conversation = example["conversation"].strip()
    analysis = example["analysis"]
    target = json.dumps(
        {
            "emotion": analysis["emotion"],
            "sentiment_polarity": analysis["sentiment_polarity"],
            "affection_score": analysis["affection_score"],
            "trend": analysis["trend"],
            "risk_level": analysis["risk_level"],
            "risk_and_subtext": analysis["risk_and_subtext"],
        },
        ensure_ascii=False,
        separators=(",", ":"),
    )
    return {
        "messages": [
            {"role": "system", "content": SYSTEM_PROMPT},
            {
                "role": "user",
                "content": f"请分析下面这段聊天记录：\n<conversation>\n{conversation}\n</conversation>",
            },
            {"role": "assistant", "content": target},
        ]
    }

converted = [convert_to_messages(x) for x in raw_data]
dataset = Dataset.from_list(converted)
dataset = dataset.remove_columns([c for c in dataset.column_names if c != "messages"])

split = dataset.train_test_split(test_size=0.05, seed=42)
train_dataset = split["train"]
eval_dataset = split["test"]

print("训练集:", train_dataset)
print("验证集:", eval_dataset)

# ================= 3. 精度自适应检查 =================
compute_capability = torch.cuda.get_device_capability(0)
use_bf16 = compute_capability[0] >= 8
if use_bf16:
    torch_dtype = torch.bfloat16
    print(f"GPU 计算能力 {compute_capability[0]}.{compute_capability[1]}，支持 BF16，使用 bf16=True")
else:
    torch_dtype = torch.float16
    print(f"GPU 计算能力 {compute_capability[0]}.{compute_capability[1]}，不支持 BF16，回退到 FP16 训练")

# ================= 4. 加载模型与分词器 =================
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME, trust_remote_code=True)
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    torch_dtype=torch_dtype,
    device_map="auto",
    trust_remote_code=True,
)
model.config.use_cache = False

# ================= 5. LoRA =================
peft_config = LoraConfig(
    r=16,
    lora_alpha=32,
    lora_dropout=0.05,
    bias="none",
    task_type="CAUSAL_LM",
    target_modules=[
        "q_proj", "k_proj", "v_proj", "o_proj",
        "gate_proj", "up_proj", "down_proj",
    ],
)

# ================= 6. 计算 warmup_steps =================
per_device_train_batch_size = 2
gradient_accumulation_steps = 8
num_train_epochs = 3
warmup_ratio = 0.03

steps_per_epoch = max(1, len(train_dataset) // per_device_train_batch_size // gradient_accumulation_steps)
total_steps = steps_per_epoch * num_train_epochs
warmup_steps = max(1, int(total_steps * warmup_ratio))

print(f"steps_per_epoch={steps_per_epoch}, total_steps={total_steps}, warmup_steps={warmup_steps}")

# ================= 7. SFTConfig（关键修复点） =================
sft_config = SFTConfig(
    output_dir=OUTPUT_DIR,

    # ---- 数据格式 ----
    packing=False,
    max_length=1024,

    # ---- 训练参数 ----
    num_train_epochs=num_train_epochs,
    per_device_train_batch_size=per_device_train_batch_size,
    per_device_eval_batch_size=2,
    gradient_accumulation_steps=gradient_accumulation_steps,
    learning_rate=1e-4,
    weight_decay=0.01,
    warmup_steps=warmup_steps,
    lr_scheduler_type="cosine",

    # ---- 🔴 精度修复：低算力 GPU 必须关掉 bf16，改用 fp16 ----
    bf16=use_bf16,
    fp16=not use_bf16,          # 当 bf16=False 时启用 fp16

    # ---- 🔴 内核修复：禁用 chunked_logprob，避免触发 addmm 的 BF16 限制 ----
    loss_type="nll",            # 默认是 "chunked_nll"，这里显式改为 "nll"

    # ---- 显存优化 ----
    gradient_checkpointing=True,

    # ---- 日志与保存 ----
    logging_steps=20,
    eval_strategy="epoch",
    save_strategy="epoch",
    save_total_limit=2,
    report_to="none",
)

# ================= 8. Trainer =================
trainer = SFTTrainer(
    model=model,
    args=sft_config,
    train_dataset=train_dataset,
    eval_dataset=eval_dataset,
    peft_config=peft_config,
    processing_class=tokenizer,
)

trainer.train()
trainer.save_model(OUTPUT_DIR)
tokenizer.save_pretrained(OUTPUT_DIR)
print("训练完成，模型已保存到:", OUTPUT_DIR)
```
